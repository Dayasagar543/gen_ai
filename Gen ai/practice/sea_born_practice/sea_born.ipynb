# seaborn

```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
sns.get_dataset_names()
```

```python
iris_flower=sns.load_dataset('iris')
iris_flower.head()
```

```python
iris_flower.shape
```

```python
# iris_flower.isnull()
```

```python
iris_flower["species"].value_counts()
```

```python
sns.scatterplot(data=iris_flower,x="sepal_length",y="sepal_width",hue="species")
```

```python
sns.scatterplot(data=iris_flower,x="petal_length",y="petal_width",hue="species")
```

```python
sns.set_style("dark")
sns.scatterplot(data=iris_flower,x="sepal_length",y="petal_length",hue="species")
```

```python
sns.set_style("whitegrid")
sns.scatterplot(data=iris_flower,x="sepal_width",y="petal_width",hue="species")
```

```python
sns.countplot(data=iris_flower,x="species")
```

```python
sns.jointplot(data=iris_flower,x="sepal_width",y="petal_width",hue="species")
```

```python
sns.set("paper")
sns.color_palette(palette='Blues')
sns.lineplot(data=iris_flower,x="sepal_width",y="petal_width",hue="species")
```

```python
sns.boxplot(data=iris_flower,x="sepal_width",y="petal_width",hue="species")
```

```python
sns.violinplot(data=iris_flower,x="sepal_width",hue="species")
```

```python
sns.violinplot(data=iris_flower,y="petal_width",hue="species")
```

```python
sns.swarmplot(data=iris_flower,x="petal_length",hue="species")
```

```python
sns.pairplot(data=iris_flower,hue="species")
```
