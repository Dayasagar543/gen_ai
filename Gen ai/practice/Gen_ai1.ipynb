# dictionaries shallow and deep copying 

```python
student_Data={'name':"Rahul",'age':26,'gender':"male"}
print(student_Data.get("name"))
print(student_Data.keys())
print(student_Data.values())
print(student_Data.items())
```

```python

for key,value in enumerate(student_Data):
    print(key,value)
    
for key,value in student_Data.items():
    print(key,value)
    
```

```python
student_1=student_Data
print(student_1)
student_Data["name"]="mahesh"
print(student_1)
print(student_Data)
```

```python
student_2=student_Data.copy()
student_Data["name"]="karthik"
print(student_2)
print(student_Data)
```

```python
nums1=[12,11,4,5,6,7,8,5,5,3]
print(nums1)
```

## conditional if and else

```python
year=int(input("please enter the year to know if its a leap year"))

if year % 4 ==0:
    if year % 100 == 0:
        if year % 400 == 0:
            print("its a leap year")
        else:
            print("not a leap year")
    else:
        print("its a leap year")
else:
    print("not a leap year")
```
